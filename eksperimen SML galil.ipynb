```python
# 1. Mencari otomatis letak folder train dan val/valid
dataset_path = "/content/fall_dataset"
train_dir = ""
val_dir = ""

for root, dirs, files in os.walk(dataset_path):
    # Mencari folder yang berakhiran train/images atau sekadar train
    if 'train' in root and ('images' in root or len(files) > 0 and files[0].endswith(('.jpg', '.png'))):
        train_dir = root
    # Mencari folder untuk validasi (val/valid)
    elif ('val' in root or 'valid' in root) and ('images' in root or len(files) > 0 and files[0].endswith(('.jpg', '.png'))):
        val_dir = root

print(f"Ketemu! Jalur Train: {train_dir}")
print(f"Ketemu! Jalur Val  : {val_dir}")

# 2. Membuat ulang file data.yaml secara otomatis dengan jalur yang benar
yaml_content = f"""
train: {train_dir}
val: {val_dir}

nc: 3
names: ['Fall Detected', 'Walking', 'Sitting']
"""

with open('/content/data.yaml', 'w') as f:
    f.write(yaml_content)

print("File data.yaml berhasil diperbarui secara otomatis dan akurat!")
```
